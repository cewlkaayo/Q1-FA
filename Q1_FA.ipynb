```python
import math

x1 = float(input("Enter number:  "))
y1 = float(input("Enter number:  "))

x2 = float(input("Enter number: "))
y2 = float(input("Enter number: "))

distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

print("Distance:", distance)
```

```python
import math

r = float(input("Enter the radius: "))

area = math.pi * r ** 2
circumference = 2 * math.pi * r

print("Area:", area)
print("Circumference:", circumference)
```

```python
import math

a = float(input("Enter leg a: "))
b = float(input("Enter leg b: "))

hypotenuse1 = math.sqrt(a ** 2 + b ** 2)
hypotenuse2 = math.hypot(a, b)

print("Hypotenuse (using math.sqrt):", hypotenuse1)
print("Hypotenuse (using math.hypot):", hypotenuse2)
```
